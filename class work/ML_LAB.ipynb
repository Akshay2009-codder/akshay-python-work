```python

age = int(input("Enter your age: "))
gender = str(input("Enter your gender(f/m): "))
if gender == "m" and 10 <= age <= 15:
    print("your fav is apple")
elif gender == "f" and 10 <= age <= 15:
    print("your fav is banana")
elif gender == "m" and 16 <= age <=20:
    print("your fav is orange")
elif gender == "f" and 16 <= age <=20:
    print("your fav is mango")
elif gender == "m" and 21 <= age <= 25:
    print("your fav is banana")
elif gender == "f" and 21 <= age <= 25:
    print("your fav is mango")
else :
    print("enter valid input")
```
